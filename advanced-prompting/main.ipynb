```python
from langchain_community.llms import Ollama

# Load Ollama model (using llama3.2 which you already have installed)
llm = Ollama(model="llama3.2")

# Test connection
print("Testing Ollama connection...")
test_response = llm.invoke("Say hello!")
print(f"Response: {test_response}")
print("✓ Ollama connected successfully!")
```

```python
# Prompt

prompt = "Give me a fun fact about cats."

output = llm.invoke(prompt)
print(output)
```

## Apply prompt template

```python
# With Ollama, templates are handled internally
# You can format messages as a string if needed:
formatted_prompt = f"User: Give me a fun fact about cats.\nAssistant:"
print(formatted_prompt)
print("\nResponse:")
print(llm.invoke(formatted_prompt))
```
